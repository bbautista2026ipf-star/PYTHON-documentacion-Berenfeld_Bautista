```python
# En Pandas existen dos estructuras de datos:
# Importamos pandas con el alias pd
import pandas as pd

# SERIES: Se componen de manera unidimensional a traves de filas (SERIES)
ventas = pd.Series([100, 200, 300, 400, 500])
ventas.index = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo']
print(ventas)

# DATAFRAMES: Se componen de manera bidimensional a traves de filas y columnas (DATAFRAMES)
ventas1 = pd.DataFrame({
    "Meses": ["Enero", "Febrero", "Marzo", "Abril"],
    "Dias": [31, 28, 31, 30]
})
ventas1.index = ["1","2", "3", "4"]

ventas1.head()
```
