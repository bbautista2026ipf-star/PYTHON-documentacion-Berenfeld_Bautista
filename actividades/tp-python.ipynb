## Actividad 4 — Condicional `if / elif / else`

Las estructuras **condicionales** permiten tomar decisiones en el programa según una condición.

**Consigna:** Solicita al usuario que ingrese su edad y muestra un mensaje según el caso:

| Condición | Mensaje |
|---|---|
| Menor de 13 años | `Sos un niño/a` |
| Entre 13 y 17 años | `Sos un adolescente` |
| 18 años o más | `Sos mayor de edad` |

**Ejemplo de salida esperada (si ingresa 15):**
```
Sos un adolescente
```

````python
# Escribí tu código acá
# Pedimos al usuario que ingrese su edad, nos aseguramos que sea un numero entero y lo guardamos en una variable
edad = int(input("Ingrese su edad: "))
# Evaluamos la edad ingresada y mostramos un mensaje correspondiente a la condicion que se cumpla
if edad >= 18:
    print("Sos mayor de edad")
elif edad < 13:
    print("Sos un niño/a")
else:
    print("Sos un adolescente")
````

## Actividad 6 — Bucle `while`: contador

El bucle **`while`** repite un bloque de código **mientras** una condición sea verdadera.

**Consigna:** Utiliza un bucle `while` para mostrar los números del 1 al 10, uno por línea.

**Ejemplo de salida esperada:**
```
1
2
3
...
10
```

````python
# Escribí tu código acá
# Definimos un contador
contador = 1
# Iniciamos el bucle
# El while evalua la condicion y mientras se cumple se seguira ejecutando
# hasta que la condicion deje de cumplirse
while contador < 11:
    print(contador)
    contador += 1
````

## Actividad 7 — Bucle `while`: acumulador

Un **acumulador** es una variable que va sumando (o acumulando) valores dentro de un bucle.

**Consigna:** Solicita al usuario que ingrese 5 números y calcula la **suma total** y el **promedio** usando un bucle `while`.

**Ejemplo de salida esperada:**
```
Suma total: 45
Promedio: 9.0
```

````python
# Escribí tu código acá
# Inicializamos el contador y el acumulador
contador = 0
Acumulador = 0
# Ejecutamos el bucle mientras el contador sea menor a 5
while contador < 5:
    # Siempre que el contador cumpla la condicion, pedimos al usuario que ingrese un numero y lo sumamos al acumulador
    numero = int(input("Ingrese un numero: "))
    # Sumamos el numero ingresado al acumulador
    Acumulador = Acumulador + numero
    # Incrementamos el contador en 1
    contador += 1

# Calculamos promedio
promedio = Acumulador / 5
# Salida
print(f"Suma total: {Acumulador}")
print(f"Promedio: {promedio}")
````

## Actividad 9 — Bucle `for`: tabla de multiplicar

**Consigna:** Solicita al usuario un número y muestra su **tabla de multiplicar** del 1 al 10 usando un bucle `for`.

**Ejemplo de salida esperada (si ingresa 7):**
```
7 x 1 = 7
7 x 2 = 14
7 x 3 = 21
...
7 x 10 = 70
```

````python
# Escribí tu código acá
# pedimos el numero entre al usuario y lo guardamos en una variable
numero = int(input("Ingrese un numero: "))
# iniciamos el bucle for
for i in range(1, 11):
    # multiplicamos el numero ingresado por el valor de i y lo guardamos en una variable
    resultado = numero * i
    # imprimimos el resultado
    print(f"{numero} x {i} = {resultado}")
````

**Consigna:** Crea un programa que:

   - El **mayor** número ingresado

**Ejemplo de salida esperada:**
```

Promedio: 32.0
```
