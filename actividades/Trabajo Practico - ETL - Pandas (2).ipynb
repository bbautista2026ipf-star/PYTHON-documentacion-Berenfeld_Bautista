````python
import pandas as pd
````

````python
# Para normalizar los valores de las ciudades podemos convertir las cadenas en minusculas y eliminar espacios
df["ciudad"] = df["ciudad"].str.lower().str.strip()
display("Esto muestra los distintos valores normalizados de la columna ciudad: ", df["ciudad"].unique())
````

````python
# Para convertir los numeros definidos como str podemos hacer lo siguiente
# Para que el programa se ejecute aunque encuentre valores que no se puedan convertir usamos errors = "coerce"
df["precio"] = pd.to_numeric(df["precio"], errors= "coerce")
display("Esto muestra los valores numericos mal definidos a enteros:", df["precio"].unique())
````

````python
### Antes de realizar reglas de negocio verificamos la existencia de datos invalidos como valores negativos
display("Este es el valor mas bajo de cantidad y precio:", df[["cantidad", "precio"]].min())
### Valores maximos
display("Este es el valor mas alto de cantidad y precio:", df[["cantidad", "precio"]].max())
````

````python
# Creamos una version del df con la siguiente condicion
df_limpio = df[(df["precio"] > 0) & (df["cantidad"] > 0)]
display("Esto muestra las reglas de negocio aplicadas:", df_limpio)
# Comprobamos que no existan valores inconsistentes
display(df_limpio["precio"].min())
display(df_limpio["precio"].max())
# Ahora con la columna cantidad
display(df_limpio["cantidad"].min())
display(df_limpio["cantidad"].max())

````

````python
# convertimos los valores originales en minusculas y quitamos espacios para detectar posibles fallas de formato
df["estado"] = df["estado"].str.lower().str.strip()
# buscamos que esos valores existan
df_buscar_valores = df[(df["estado"] == "pendiente") | (df["estado"] == "enviado") | (df["estado"] == "entregado")]
display("Esto comprueba que existan esos valores en la categoria:", df_buscar_valores["estado"].nunique())
````

````python
# Convertimos la columna de fecha al formato solicitado
df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d", errors="coerce")
# fechas invalidas (nat)
df_fecha_invalida = df["fecha"].isna()
# Fechas con el formato solicitado y nat
display("Esto muestra el formato indicado para las fechas y debajo aquellos que no sigan el formato:", df["fecha"], df_fecha_invalida)
````

````python
# Se eliminan campos nulos o innecesarios
df = df.drop(columns=["comentario_interno"])
# # Se verifica su inexistencia
display(df)
````

````python
# Identificamos valores duplicados
display("Esto recopila datos duplicados:", df[df.duplicated()])
````

````python
# Eliminamos datos duplicados
df = df.drop_duplicates()
# Verificamos
display("Esto verifica la inexistencia de datos duplicados", df)
````
